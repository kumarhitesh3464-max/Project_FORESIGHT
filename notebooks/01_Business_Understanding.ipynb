# Project FORESIGHT

## Phase 1: Business Understanding

### Project Overview

Project FORESIGHT is an end-to-end retail demand forecasting and inventory optimization project.

The project focuses on predicting future product demand, identifying inventory risks, and building an interactive dashboard to support business decision-making.

The complete workflow includes business understanding, data collection, data cleaning, exploratory data analysis, feature engineering, demand forecasting, model evaluation, risk scoring, dashboard development, deployment, and executive reporting.

# Business Problem

Retail businesses often face inventory-related challenges such as:

- Stockouts leading to lost sales
- Overstock resulting in high holding costs
- Poor demand forecasting
- Inefficient replenishment planning
- Revenue loss due to inaccurate inventory decisions

The organization requires a forecasting solution that predicts future demand and identifies inventory risks before they occur.

# Business Objectives

The main objectives of Project FORESIGHT are:

- Understand historical sales behavior.
- Forecast future product demand.
- Improve inventory planning.
- Detect stockout and overstock risks.
- Provide actionable business recommendations.
- Build an interactive Streamlit dashboard.
- Deploy the complete forecasting solution.

# Available Datasets

The project uses the following datasets:

1. sales_transaction.csv
   - Daily sales transactions
   - Customer purchases
   - Quantity sold
   - Revenue
   - Discounts

2. sku_master.csv
   - Product information
   - Category
   - Brand
   - Product pricing

3. inventory_snapshot.csv
   - Current inventory
   - Reorder point
   - Safety stock
   - Restock information

4. calendar.csv
   - Generated from the sales transaction date column
   - Used for feature engineering and time-series analysis
