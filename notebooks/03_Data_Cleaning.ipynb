# Phase 2: Data Cleaning

## Objective

The objective of this notebook is to clean the datasets before performing analysis and forecasting.

Tasks:

- Check missing values
- Check duplicate records
- Verify data types
- Convert date columns
- Remove duplicate records (if any)
- Validate numerical columns

```python
# Import Libraries

import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)

print("Libraries Imported Successfully")
```

```python
import pandas as pd

sales = pd.read_csv("../data/raw/sales_transactions.csv")

sku = pd.read_csv("../data/raw/sku_master.csv")

inventory = pd.read_csv("../data/raw/inventory_snapshot.csv")
```

# Missing Values Analysis

```python
print("Sales Missing Values")

sales.isnull().sum()
```

```python
print("SKU Missing Values")

sku.isnull().sum()
```

```python
print("Inventory Missing Values")

inventory.isnull().sum()
```

# Duplicate Records

```python
print("Sales Duplicate Records")

print(sales.duplicated().sum())
```

```python
print("SKU Duplicate Records")

print(sku.duplicated().sum())
```

```python
print("Inventory Duplicate Records")

print(inventory.duplicated().sum())
```

# Remove Duplicate Records

```python
sales = sales.drop_duplicates()

sku = sku.drop_duplicates()

inventory = inventory.drop_duplicates()

print("Duplicate Records Removed")
```

# Data Types Verification

```python
sales.dtypes
```

```python
sku.dtypes
```

```python
inventory.dtypes
```

# Convert Date Columns

```python
sales["date"] = pd.to_datetime(sales["date"])

inventory["last_restock_date"] = pd.to_datetime(
    inventory["last_restock_date"]
)

print("Date Columns Converted Successfully")
```

# Verify Converted Data Types

```python
sales.dtypes
```

```python
inventory.dtypes
```

# Check Numeric Columns

```python
sales.describe()
```

```python
inventory.describe()
```

# Data Quality Summary

Cleaning activities completed:

- Checked missing values
- Checked duplicate records
- Removed duplicate rows
- Converted date columns
- Verified numerical columns

The datasets are now ready for Exploratory Data Analysis (EDA).

```python
# =====================================================
# CREATE GITHUB FRIENDLY SAMPLE DATASETS
# =====================================================

import pandas as pd

# Sample only sales dataset
sales_sample = sales.sample(
    n=100000,
    random_state=42
)

# SKU & Inventory already comparatively small
sku_sample = sku.copy()
inventory_sample = inventory.copy()

# Save processed datasets

sales_sample.to_csv(
    "../data/processed/sales_clean.csv",
    index=False
)

sku_sample.to_csv(
    "../data/processed/sku_clean.csv",
    index=False
)

inventory_sample.to_csv(
    "../data/processed/inventory_clean.csv",
    index=False
)

print("="*60)
print("✅ Sample datasets created successfully")
print("="*60)

print("Sales Rows :", len(sales_sample))
print("SKU Rows :", len(sku_sample))
print("Inventory Rows :", len(inventory_sample))
```
