# Phase 7: Streamlit Dashboard

## Objective

The objective of this phase is to develop an interactive dashboard for retail demand forecasting and inventory risk monitoring.

The dashboard enables business users to monitor sales trends, demand forecasts, inventory levels, and product risk categories through an easy-to-use interface.

---

## Dashboard Features

- Business KPIs
- Demand Forecast
- Inventory Overview
- Risk Distribution
- Category Analysis
- Store Analysis
- Interactive Filters
- Download Processed Dataset
